```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

data = 'https://raw.githubusercontent.com/Alam-A99/DATAAKSI/main/dataset_fiskal_97_hipoFull.csv'
df = pd.read_csv(data, header=None)
df = pd.DataFrame({
    'Kota'  : df.iloc[2:, 0],
    'Tahun' : df.iloc[2:, 1],
    'X1'    : df.iloc[2:, 11],
    'X2'    : df.iloc[2:, 16],
    'Y'     : df.iloc[2:, 20],
})

df = df[df['Kota'] == 'Kota Banda Aceh'].reset_index(drop=True)
for col in ['X1', 'X2', 'Y']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
print(df)

x = sm.add_constant(df[['X1', 'X2']])
y = df['Y']
model = sm.OLS(y, x)
results = model.fit()
print(results.summary())
```
